## Find whether the entered number is palindrome or not

```python
num = int(input("Enter a number: "))
absNum = abs(num)

rev = absNum % 10
absNum = absNum // 10
while(absNum > 0):
  r = absNum % 10
  absNum = absNum // 10
  rev = rev * 10 + r
if(num < 0):
  rev = rev - 2 * rev
if(num == rev):
  print("Palindrome")
else:
  print("Not a Palindrome")
```

## for loop tutorial

```python
# factoril using while loop

num = int(input("Enter a Number: "))
fact = 1
if (num < 0):
  print("Not Defined")
else:
  while(num>0):
    fact = fact * num
    num = num - 1
  print(fact)
```

```python
# factorial using for loop

num = int(input("Enter a Number: "))
fact = 1

if (num<0):
  print("Not Defined")
else:
  for i in range(num,1, -1):
    fact = fact * i
  print(fact)
```
