## Write a piece of code that let you attempt as many time till you are correct.

```python
print("When did India got its independence?")
year = int(input())

while(year!=1947):
  print("You got this wrong. Enter once again")
  year = int(input())
print("Waow, you got it right!")
```

## Write a code to write factorial of a number using while loop

```python
print("Enter a number")
n = int(input())

i = 1
answer = 1
while(i<=n):
  answer = answer*i
  i = i+1
print(answer)
```

```python
num = int(input("Enter a number: "))

fact = 1

while(num > 0):
  fact = fact*num
  num = num - 1

print(fact)
```

```python
# to incorporate negative number

num = int(input("Enter a number: "))

fact = 1

if (num< 0):
  print("Factorial not defined")
else:
  while(num>0):
    fact = fact*num
    num = num - 1
  print(fact)
```

## Find the number of digits in a given number

```python
num = abs(int(input("Enter a number: ")))
digits = 1

while(num>9):
  num = num//10
  digits = digits + 1
print(digits)
```

## Reverse the digits in the given number

```python
num = int(input("Enter a number: "))
rev = num % 10
num = num // 10
while(num>0):
  r = num % 10
  num = num // 10
  rev = rev * 10 + r
print(rev)
```
